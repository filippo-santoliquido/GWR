# A normalizing flow in one dimension

A normalizing flow turns a simple probability distribution into a complicated one.
It does so with an invertible function. This notebook shows the idea with one variable.

1. We draw random numbers $u$ from a standard Gaussian, the *base distribution* $p_0(u)$.
2. We apply an invertible function, $\theta = f(u)$.
3. The numbers $\theta$ follow a new distribution $q(\theta)$, given by the change-of-variables formula

$$q(\theta) = p_0(u)\,\left|\frac{\mathrm{d}u}{\mathrm{d}\theta}\right|, \qquad u = f^{-1}(\theta).$$

In a real flow the function $f$ is learned by a neural network.
Here we build it by hand, so that $q(\theta)$ has two peaks.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

from gwr_style import COLORS, TEXTWIDTH, save_figure

rng = np.random.default_rng(1)  # random number generator with a fixed seed
```

## The distribution we want to obtain

Our target is the sum of two Gaussians: one centred at $-1.2$ and one at $1.6$.

```python
def target_pdf(theta):
    """Probability density of the target distribution."""
    return 0.65 * norm.pdf(theta, -1.2, 0.45) + 0.35 * norm.pdf(theta, 1.6, 0.6)


def target_cdf(theta):
    """Cumulative distribution of the target distribution."""
    return 0.65 * norm.cdf(theta, -1.2, 0.45) + 0.35 * norm.cdf(theta, 1.6, 0.6)
```

## The invertible function

In one dimension there is a simple recipe to send one distribution onto another.
Match the cumulative distributions: the point $u$ that has 30% of the Gaussian below it
is sent to the point $\theta$ that has 30% of the target below it, and so on.
We tabulate this function on a fine grid.

```python
theta_grid = np.linspace(-4, 5, 4000)

# For every theta on the grid, the value of u with the same cumulative probability
u_of_theta = norm.ppf(np.clip(target_cdf(theta_grid), 1e-12, 1 - 1e-12))


def flow(u):
    """The invertible function theta = f(u), obtained by interpolating the table."""
    return np.interp(u, u_of_theta, theta_grid)


# Step 1 and 2: draw Gaussian random numbers and transform them
u_samples = rng.standard_normal(20000)
theta_samples = flow(u_samples)

# Step 3: the density of theta from the change-of-variables formula
du_dtheta = np.gradient(u_of_theta, theta_grid)
q_theta = norm.pdf(u_of_theta) * np.abs(du_dtheta)
```

## The figure

Left: the base distribution. Centre: the function $f$. Right: the transformed distribution.
In the right panel the histogram shows the transformed samples, the solid line is the formula above,
and the dashed line is the target. They agree.

```python
fig, axes = plt.subplots(1, 3, figsize=(TEXTWIDTH, 0.33 * TEXTWIDTH), gridspec_kw={"wspace": 0.42})
u = np.linspace(-3.5, 3.5, 300)

# Left: base distribution
ax = axes[0]
ax.hist(u_samples, bins=60, density=True, color=COLORS["lightblue"], alpha=0.7)
ax.plot(u, norm.pdf(u), color=COLORS["blue"])
ax.set_xlabel("$u$")
ax.set_ylabel("$p_0(u)$", labelpad=1)
ax.set_title("Base distribution")
ax.set_xlim(-3.5, 3.5)

# Centre: the invertible function, with three example points
ax = axes[1]
ax.plot(u, flow(u), color=COLORS["teal"])
for u0 in (-1.5, 0.0, 1.5):
    ax.plot([u0, u0, -3.5], [-3.2, flow(u0), flow(u0)], color=COLORS["grey"], lw=0.5, ls=":")
    ax.plot(u0, flow(u0), "o", ms=2.5, color=COLORS["teal"])
ax.set_xlabel("$u$")
ax.set_ylabel(r"$\theta=f_\phi(u)$", labelpad=-3)
ax.set_title("Invertible map")
ax.set_xlim(-3.5, 3.5)
ax.set_ylim(-3.2, 3.8)

# Right: transformed distribution
ax = axes[2]
ax.hist(theta_samples, bins=80, density=True, color=COLORS["orange"], alpha=0.45)
ax.plot(theta_grid, q_theta, color=COLORS["orange"])
ax.plot(theta_grid, target_pdf(theta_grid), color="k", lw=0.6, ls="--")
ax.set_xlabel(r"$\theta$")
ax.set_ylabel(r"$q_\phi(\theta)$", labelpad=1)
ax.set_title("Transformed distribution")
ax.set_xlim(-3.2, 3.8)

for ax in axes:
    ax.tick_params(length=2)

save_figure(fig, "fig_normalizing_flow.pdf")
plt.show()
```
